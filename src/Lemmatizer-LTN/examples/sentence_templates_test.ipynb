# Latin Sentence Templates — Lemmatizer Test

Tests the lemmatizer against example sentences from the Latin Sentence Templates (MASTER) document.
For each sentence, we look up every word's lemma and verify the morphological forms exist in the database.

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path().resolve().parent))
from latin_lemmatizer import get_lemmas, get_forms, LatinLemmatizer

print("[OK] Imported latin_lemmatizer")
```

```python
def analyze_sentence(sentence: str, translation: str = ""):
    """Analyze each word in a Latin sentence using the lemmatizer."""
    print(f"Sentence: {sentence}")
    if translation:
        print(f"Translation: {translation}")
    print()
    
    words = [w.strip(".,!?;:") for w in sentence.split() if w.strip(".,!?;:")]
    all_found = True
    
    for word in words:
        lemma = get_lemmas(word, top=True)
        if lemma:
            print(f"  {word:20} -> lemma: {lemma['lemma_diac']:15} ({lemma['pos']})")
        else:
            print(f"  {word:20} -> NOT FOUND")
            all_found = False
    
    status = "PASS" if all_found else "FAIL (missing words)"
    print(f"\n  Result: {status}")
    print("-" * 70)
    return all_found
```

## Accusative Object Sentences

```python
sentences = [
    ("Femina virum laudat", "The woman praises the man."),
    ("Poeta pecuniam et famam habet", "The poet has money and fame."),
]

for s, t in sentences:
    analyze_sentence(s, t)
    print()
```

## Dative Indirect Object Sentences

```python
sentences = [
    ("Agricola agro aquam adfert", "The farmer brings water to the field."),
    ("Magister discipulo scientiam et sapientiam dat", "The teacher gives knowledge and wisdom to the students."),
]

for s, t in sentences:
    analyze_sentence(s, t)
    print()
```

## Genitive of Possession

```python
sentences = [
    ("Arbores silvae", "The trees of the forest."),
    ("Rex cives urbis videt", "The king sees the city's citizens."),
]

for s, t in sentences:
    analyze_sentence(s, t)
    print()
```

## Ablative (Adverbial / Means)

```python
sentences = [
    ("Miles gladio vincit", "The soldier conquers by the sword."),
    ("Poeta carmina stilo scribit", "The poet writes poems with a stylus."),
    ("Filius agricolae rure venit", "The farmer's son is coming from the countryside."),
    ("Miles gladio pugnat", "The soldier fights with a sword."),
]

for s, t in sentences:
    analyze_sentence(s, t)
    print()
```
